```python
import time
import os
from MCTS_NEW import *
import pickle

code_begins = time.time()

# print(device_lib.list_local_devices())

os.environ['CUDA_VISIBLE_DEVICES'] = '0'
# os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'
args = {
    "Total_time": 100,          # Time window available for scheduling in seconds
    "NumTasks": 5,             # Number of tasks
    "NumFeatures": 8,           # start, interval, tardiness
    "ValidationSize": 20,     # Number of problems per iteration
    "TestSize": 100,
    "max_cost": 10**10,            # Maximum cost used to normalize all costs
    "NumMCTSRollouts": 256,
}

nmcts = MCTS(args)

optimal_avg_cost, est_avg_cost, edt_avg_cost, optimal_nodes, optimal_avg_dropped, est_avg_dropped, edt_avg_dropped\
    = nmcts.compare()

print('      Optimal AvgCost: %.2f' % optimal_avg_cost)
print('      EST AvgCost: %.2f' % est_avg_cost)
print('      EDT AvgCost: %.2f' % edt_avg_cost)
print('      Optimal Dropped: %.2f' % optimal_avg_dropped)
print('      EST Dropped: %.2f' % est_avg_dropped)
print('      EDT Dropped: %.2f' % edt_avg_dropped)
# print('      MCTS AvgCost: %.2f' % mcts_avg_cost)

folder = 'results1'
if not os.path.exists(folder):
    os.mkdir(folder)

filename = 'optimal_cost'
filepath = os.path.join(folder, filename)
np.save(filepath, optimal_avg_cost)

filename = 'est_cost'
filepath = os.path.join(folder, filename)
np.save(filepath,est_avg_cost)

filename = 'edt_cost'
filepath = os.path.join(folder, filename)
np.save(filepath,edt_avg_cost)

filename = 'optimal_dropped'
filepath = os.path.join(folder, filename)
np.save(filepath,optimal_avg_dropped)

filename = 'est_dropped'
filepath = os.path.join(folder, filename)
np.save(filepath,est_avg_dropped)

filename = 'edt_dropped'
filepath = os.path.join(folder, filename)
np.save(filepath,edt_avg_dropped)

filename = 'optimal_nodes'
filepath = os.path.join(folder, filename)
f = open(filepath, 'wb')
pickle.dump(optimal_nodes, f)

# Print Runtime
time=time.time() - code_begins
hours=int(time/3600)
minutes=int((time - 3600*hours)/60)
seconds=int(time - 3600*hours - 60*minutes)
print("--- %s hrs, %s mins, %s secs ---" % (hours, minutes, seconds))

```
